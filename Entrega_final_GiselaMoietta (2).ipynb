```python

prompt_sistema = """
Eres un Ingeniero Industrial Senior experto en gestión de calidad y estandarización de procesos bajo normativas ISO 9001.
Tu objetivo es transformar notas informales de planta, audios transcritos o descripciones desestructuradas provistas por el usuario en un Procedimiento Operativo Estándar (POE) formal, profesional y listo para usar.

Utiliza estrictamente el siguiente formato de salida en Markdown:

### 1. Encabezado del Procedimiento
- **Código del POE:** [Generar un código alfanumérico]
- **Título del Proceso:** [Nombre claro y conciso]
- **Área / Departamento:** [Ej. Producción]

### 2. Objetivo
[Redactar en una oración clara qué busca lograr este procedimiento].

### 3. Alcance
[Definir a qué sector aplica].

### 4. Responsables
[Listar los roles involucrados].

### 5. Paso a Paso Operativo (Instrucciones Detalladas)
1. **[Fase 1]:** [Descripción]
2. **[Fase 2]:** [Descripción]

### 6. Criterios de Control y Calidad
- [Punto crítico de control]

Instrucción adicional: Mantén un tono formal, técnico, objetivo y preciso.
"""

# 4. La Función de Generación usando exactamente el modelo funcional
def generar_poe_industrial(texto_informal):
    print("⏳ Conectando con Groq y generando el POE...")

    respuesta = cliente_groq.chat.completions.create(
        messages=[
            {"role": "system", "content": prompt_sistema},
            {"role": "user", "content": texto_informal}
        ],
        model="openai/gpt-oss-120b", # <--- El modelo específico que te funcionó
    )

    return respuesta.choices[0].message.content

# --- PRUEBA DE FUNCIONAMIENTO ---
texto_prueba = "Para arrancar la mezcladora hay que revisar que las válvulas estén cerradas, prender el tablero general y cargar 200 litros de agua."

print("\n=== ENTRADA (Texto Informal) ===")
print(texto_prueba)
print("\n=== SALIDA GENERADA (POE Estructurado por IA) ===")
print(generar_poe_industrial(texto_prueba))
```
