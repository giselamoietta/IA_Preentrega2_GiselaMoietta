```python
import time

def generar_poe_industrial(texto_informal_proceso):
    """
    Función de Prueba de Concepto (POC) que simula el procesamiento
    de un texto informal de planta y genera un POE estructurado.
    """
    print("🤖 Conectando con el motor de Prompt Engineering (Modo Simulado)...")
    time.sleep(1) # Simula el tiempo de respuesta de la API

    # Análisis básico del texto ingresado para darle dinamismo a la simulación
    texto_lower = texto_informal_proceso.lower()

    # Generación dinámica del POE basada en el texto del usuario
    poe_resultado = f"""
### 1. Encabezado del Procedimiento
- **Código del POE:** POE-IND-2026-001
- **Título del Proceso:** Estandarización de Operación de Planta y Mezcla
- **Área / Departamento:** Producción y Calidad

### 2. Objetivo
Estandarizar de forma segura y eficiente la ejecución operativa del proceso industrial ingresado, minimizando desvíos y asegurando los estándares de calidad establecidos.

### 3. Alcance
Aplica a todo el personal operativo, de planta y supervisores encargados de la ejecución y control de esta línea de producción.

### 4. Responsables
- Operario de Línea / Producción
- Supervisor de Turno

### 5. Paso a Paso Operativo (Instrucciones Detalladas)
1. **Verificación Inicial:** Inspeccionar visualmente el área de trabajo y constatar el estado óptimo de las válvulas y tableros principales.
2. **Ejecución del Proceso:** Proceder con la carga o activación de insumos según los parámetros relevados: *"{texto_informal_proceso}"*.
3. **Control Intermedio:** Monitorear de forma continua las variables críticas de proceso (presión, temperatura o pH según corresponda).

### 6. Criterios de Control y Calidad
- Control estricto de parámetros al inicio y cierre de campaña.
- Notificación inmediata al Supervisor ante cualquier desvío detectado fuera de rango.
"""
    return poe_resultado

# --- PRUEBA DE FUNCIONAMIENTO EN EL NOTEBOOK ---
texto_prueba = "Para arrancar la mezcladora hay que revisar que las válvulas estén cerradas, prender el tablero general y cargar 200 litros de agua."

print("=== ENTRADA (Texto Informal) ===")
print(texto_prueba)
print("\n=== SALIDA GENERADA (POE Estructurado) ===")
print(generar_poe_industrial(texto_prueba))
```
